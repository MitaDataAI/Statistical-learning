```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
```

# a) When Y-> X = X-> Y

A) The coefficient estimate of (X) onto (Y):

$$
\hat{B}=\frac{\sum_{i=1}^{n}(x_i y_i)}{\sum_{j=1}^{n} x_j^2}
$$

The coefficient estimate of (Y) onto (X):

$$
\hat{B}=\frac{\sum_{i=1}^{n}(x_i y_i)}{\sum_{j=1}^{n} y_j^2}
$$

In order to have the same coefficient for both, we must have:

$$
\sum_{j=1}^{n} y_j^2=\sum_{j=1}^{n} x_j^2
$$

# b) Y -> X # X -> Y

```python
rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = x * 3
```

```python
y_squared = y**2
y_squared_sum = np.sum(y_squared)
y_squared_sum
```

```python
x_squared =  x**2
x_squared_sum = np.sum(x_squared)
x_squared_sum
```

On voit que y² et x² sont différents.

```python
# Correction
df = pd.DataFrame()
df['X'] = rng.normal(size=100)
df['Y'] = x * 3
print(sum(df['X']**2))
print(sum(df['Y']**2))
```

```python
model_1 = sm.OLS(df['Y'], df['X']).fit()
print(model_1.params)
```

```python
model_2 = sm.OLS(df["X"], df["Y"]).fit()
print(model_2.params)
```

# c) Y -> X = X -> Y

```python
rng = np.random.default_rng()
df = pd.DataFrame()

df['X'] = rng.normal(size=100)
df['Y'] = df['X']

print((df['X']**2).sum())
print((df['Y']**2).sum())
```

```python
model_1 = sm.OLS(df['Y'], df['X']).fit()
print(model_1.params)
```

```python
model_2 = sm.OLS(df["X"], df["Y"]).fit()
print(model_2.params)
```
