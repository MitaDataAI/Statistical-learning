# Package

```python
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

# Génération des données

```python
rng = np.random.default_rng(1)
x = rng.normal(size = 100)
y = 2 * x + rng.normal(size = 100)
```

# Scatter

```python
plt.scatter(x,y)
plt.show()
```

# a) LR y onto x without intercept

```python
model = sm.OLS(y, x).fit()
```

```python
print("Beta :", model.params)
print("t-statistic:", model.tvalues)
print("p-values:", model.pvalues)
```

## Commentaire
- x explique très significativement y. En effet, p_value < 0.01.
- Si x augmente d'une unité, en moyenne, y augmente de 1.97 unités. 

# b) 

```python
SE = np.sqrt(model.mse_resid)
print(SE)
```
