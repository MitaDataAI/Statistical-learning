# Package

```python
from ISLP import load_data
import statsmodels.formula.api as smf
import statsmodels.api as sm
from matplotlib.pyplot import subplots, show, scatter
import pandas as pd
import seaborn as sns
```

# Importation des données

```python
Boston_df = load_data("Boston")
```

```python
Boston_df.head()
```

* **`crim`** : taux de criminalité par habitant, par ville.

* **`zn`** : proportion de terrains résidentiels zonés pour des lots de plus de 25 000 pieds carrés.

* **`indus`** : proportion de surfaces consacrées aux activités industrielles (hors commerce de détail) par ville.

* **`chas`** : variable indicatrice liée à la rivière Charles (= 1 si le secteur borde la rivière, 0 sinon).

* **`nox`** : concentration d’oxydes d’azote (en parties par 10 millions).

* **`rm`** : nombre moyen de pièces par logement.

* **`age`** : proportion de logements occupés par leur propriétaire construits avant 1940.

* **`dis`** : moyenne pondérée des distances aux cinq principaux centres d’emploi de Boston.

* **`rad`** : indice d’accessibilité aux axes routiers radiaux.

* **`tax`** : taux d’imposition foncière (valeur totale) pour 10 000 $.

* **`ptratio`** : ratio élèves / enseignants par ville.

* **`black`** : variable calculée selon la formule 1000 (Bk − 0,63)², où Bk est la proportion de population noire par ville.

* **`lstat`** : proportion de la population appartenant aux catégories socio-économiques défavorisées (en pourcentage).

* **`medv`** : valeur médiane des logements occupés par leur propriétaire, en milliers de dollars.

# b) Multiple Regression

```python
Boston_df["intercept"] = 1
```

```python
model = sm.OLS(Boston_df["crim"], Boston_df.drop("crim", axis = 1)).fit()
print(model.summary())
```

```python
model.params
```

## Commentaire

### Significativité globale : 
Le p-value du test de Fisher est inférieur à 0.01. Donc, le modèle est très significatif au niveau global. 

### Significativité des variables : 
Seules les variables suivantes sont significatives : 
- dis : distance par rapport au centre ville ; 
- rad : indice d'accessibilité aux routes radieuses ; 
- medv : La valeur médian des maisons du cartier ; 

### Relation avec le crime
- dis : -1 : Plus l'endroit est éloigné des centres d'emplois de Boston, le crime chutte doublement. 
- rad : 0.6 : plus la localité est proche d'une route radieuse, alors le taux de criminalité augmente de 60%. 
- medv : -0.22 : si la valeur médian des maisons augmente de mille dollars, alors le taux de criminalité diminue de 22% en moyenne.  

# c) a VERSUS b

Ce qui m'a marqué c'est que le modèle multivarié (b) a juste 03 vairables significatives alors que dans la régression linéaire simple (a), tout est significatif (sauf chas). 

```python
# Création d'un dataframe
model_df_analysis = pd.DataFrame()
model_df_analysis["coef_mult_OLS"] = model.params
model_df_analysis.head()
```

```python
coeff_LR_simple = pd.DataFrame(
    {
        "coef_simple_OLS": [
            model_age.params.iloc[1],
            model_dis.params.iloc[1],
            model_chas.params.iloc[1],
            model_indus.params.iloc[1],
            model_lstat.params.iloc[1],
            model_medv.params.iloc[1],
            model_nox.params.iloc[1],
            model_ptratio.params.iloc[1],
            model_rad.params.iloc[1],
            model_rm.params.iloc[1],
            model_tax.params.iloc[1],
            model_zn.params.iloc[1],
        ]
    },
    index=[
        "age", "dis", "chas", "indus", "lstat",
        "medv", "nox", "ptratio", "rad", "rm",
        "tax", "zn"
    ]
)

coeff_LR_simple.head()
```

```python
model_df_analysis = model_df_analysis.join(coeff_LR_simple)
model_df_analysis.head()
```
