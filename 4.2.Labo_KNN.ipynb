# Package

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from ISLP import load_data
from ISLP.models import ModelSpec as MS
from ISLP import confusion_table
from sklearn.preprocessing import StandardScaler
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
```

# Dataset

```python
Smarket = load_data('Smarket')
Smarket.drop("Direction", axis=1).corr()
Smarket.head()
```

```python
# Dataframe
allvars = Smarket.columns[1:3]
design = MS(allvars)
X = design.fit_transform(Smarket)
X.head()
```

```python
# Train and Test
train = (Smarket.Year < 2005)
X_train, X_test = X.loc[train], X.loc[~train]
```

```python
# Cible
D = Smarket.Direction
L_train, L_test = D.loc[train], D.loc[~train]
```

# Exercice 1

```python
# instancier la classe de KNN 
knn1 = KNeighborsClassifier(n_neighbors=1)

# Entrainer le KNN
knn1.fit(X_train, L_train)

# Prédiction de nouvelle donnée X_test avec KNN 
knn_pred = knn1.predict(X_test)

# Evaluation du KNN 
confusion_table(knn_pred, L_test)
```

```python
print("La bonne prédiction est de :", ((43 + 83) / (43 + 68 + 58 +83)) * 100, "%")
```

Ce n'est pas une bonne prédiction. 

```python
# KNN avec K = 3
## Instancier KNN
knn3 = KNeighborsClassifier(n_neighbors = 3)

## Entrainement du KNN
knn3.fit(X_train, L_train)

## Prédiction avec le KNN 
knn3_pred = knn3.predict(X_test)

## Evaluation
CT = confusion_table(knn3_pred, L_test)
```

```python
CT
```

```python
correct_predictions = CT.at["Down", "Down"] + CT.at["Up", "Up"]
total_predictions = CT.values.sum()
accuracy = (correct_predictions / total_predictions) * 100
accuracy
```

# Exercice 2

```python
Caravan = load_data("Caravan")
print(Caravan.shape)
print(Caravan.head())
```

```python
Caravan = load_data('Caravan')
Purchase = Caravan.Purchase
Purchase.value_counts()
```

```python
feature_df = Caravan.drop(columns="Purchase")
```

```python
# Standartiser
scaler = StandardScaler(with_mean = True, # Moyenne égale à 0
                        with_std = True, #standard deviation égale à 1
                        copy = True) #utilisation du df
```

```python
scaler.fit(feature_df)
X_std = scaler.transform(feature_df)
```

```python
feature_std = pd.DataFrame(
    X_std, 
    columns= feature_df.columns
)

feature_std.describe()
```

```python
# Split
(X_train,
 X_test,
 y_train,
 y_test) = train_test_split(feature_std, 
                            Purchase,
                            test_size=1000,
                            random_state=0)
```

```python
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1.fit(X_train, y_train)
knn1_pred = knn1.predict(X_test)
# np.mean converti le booléen de True en 1 et False en 0
np.mean(y_test != knn1_pred), np.mean(y_test != "No")
```

Le taux d'erreur est dans 1000 observations est de 11%. 
Le modèle ne performe pas puisque le taux d'achat est estimé à 6,7%. 

çà veut dire que si on essaye de vendre l'assurance aléatoirement au client, seul 6% qui achètent le produit. C'est pas intéressant. 

Le plus important est d'identifier correctement les acheteurs. 

```python
confusion_table(knn1_pred, y_test)
```

Dans 62 clients (53 + 9), ou 14,6% vont acheter l'assurance. C'est le double de ce qu'on obtient à partir d'un random model (6,7%). 

# Tuning Parameters

```python
for K in range(1,6):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_table(knn_pred, y_test)
    templ = ('K={0:d}: # predicted to rent: {1:>2},' + ' # who did rent {2:d}, accuracy {3:.1%}')
    pred = C.loc['Yes'].sum()
    did_rent = C.loc['Yes', 'Yes']
    print(templ.format(
        K,
        pred,
        did_rent,
        did_rent / pred
    ))
```

# Comparison to Logistic Regression

```python
logit = LogisticRegression(C = 1e10, solver = 'liblinear')
logit.fit(X_train, y_train)
logit_pred = logit.predict_proba(X_test)
logit_labels = np.where(logit_pred[:,1] > 5, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```

```python
logit_labels = np.where(logit_pred[:, 1] > 0.25, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```
